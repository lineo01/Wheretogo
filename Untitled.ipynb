```python
#api for altitude of the locations 

import pandas as pd
import requests
import numpy as np
import time

def elevation(lat, lon):
        url = "https://api.open-meteo.com/v1/elevation"
        lat_ = ",".join(map(str, lat))
        lon_ = ",".join(map(str, lon))
        params = {"latitude": lat_, "longitude": lon_}
        try:
            response = requests.get(url, params=params)
            if response.status_code == 200:
                data = response.json()
                return data["elevation"]
        except Exception as e:
            print(f"Error fetching elevation for ({lat}, {lon}): {e}")
        return None
initlat=26.0
initlong=85.5

rivlat=26.5
rivlong=86.5
for i in range(10):
    _latitudes = [float(round(x, 4)) for x in list(np.linspace(initlat, initlat+0.15, 90))]
    _longitudes = [float(round(x, 4)) for x in list(np.linspace(initlong, initlong+0.15, 90))]
    _lat = [float(round(x, 4)) for x in list(np.linspace(rivlat, rivlat+0.15, 90))]
    _long = [float(round(x, 4)) for x in list(np.linspace(rivlong, rivlong+0.15, 90))]
    
    initlat += 0.15
    initlong += 0.15
    rivlat += 0.15
    rivlong += 0.15
    elev_a = elevation(_latitudes, _longitudes)
    elev_b = elevation(_lat, _long)

    if elev_a is not None and elev_b is not None:
        a = np.array(elev_a)
        b = np.array(elev_b)
        print(a - b)
    else:
        print("Skipping math for this iteration due to API failure.")
        
    time.sleep(1)
```
